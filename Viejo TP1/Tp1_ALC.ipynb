```python
def leer_archivo(input_file_path):

    f = open(input_file_path, 'r')
    n = int(f.readline())
    m = int(f.readline())
    W = np.zeros(shape=(n,n))
    for _ in range(m):
    	line = f.readline()
    	i = int(line.split()[0]) - 1
    	j = int(line.split()[1]) - 1
    	W[j,i] = 1.0
    f.close()

    return W
```

```python
def dibujarGrafo(W, print_ejes=True):

    options = {
    'node_color': 'yellow',
    'node_size': 200,
    'width': 3,
    'arrowstyle': '-|>',
    'arrowsize': 10,
    'with_labels' : True}

    N = W.shape[0]
    G = nx.DiGraph(W.T)

    #renombro nodos de 1 a N
    G = nx.relabel_nodes(G, {i:i+1 for i in range(N)})
    if print_ejes:
        print('Ejes: ', [e for e in G.edges])

    nx.draw(G, pos=nx.spring_layout(G), **options)

```

```python
def vectorGrado(M):
    vectorGrado=np.zeros(shape=(len(M)))
    for m in range(0,len(M),1): # columna m
       for n in range(0,len(M),1): #fila n
           if M[n][m]==1:
              vectorGrado[m]=vectorGrado[m]+1
    return vectorGrado

def matrizPuntajes(M, C):
  matrizPuntajes = np.zeros([len(M), len(M)])
  for m in range(0,len(M),1): # columna m
       for n in range(0,len(M),1): #fila n
          if M[n][m]==1:
            if C[m] != 0:
              matrizPuntajes[n][m] = 1/C[m]
  return matrizPuntajes

def matrizA(M, C, P):
  matrizA = np.zeros([len(M), len(M)])
  for m in range(0,len(M),1): # columna m
       for n in range(0,len(M),1): #fila n
          if C[m] != 0:
            matrizA[n][m] = (1-P)/len(M)+(P * M[n][m])/C[m]
          else:
            matrizA[n][m] = 1/len(M)
  return matrizA
```

```python
def pivoteo_partial(A):
    P = np.identity(len(A))
    for j in range(0,len(A)): #recorre columna
        max_col = 0
        pos_fila_max = j #indice de la columna actual
        for i in range(j, len(A)): #recorre fila
            if np.abs(A[i][j]) > max_col:
                max_col = np.abs(A[i][j])
                pos_fila_max = i

        P[[j, pos_fila_max]] = P[[pos_fila_max,j]]
        A = P @ A
    return P

A=[[1, -8, 1],[-4, -2, 1],[1, -4, 1]]
pivoteo_partial(A)
```

```python
def resolver_PA_LU(A):
    U = np.copy(A)
    T= np.identity(len(A))
    L= np.identity(len(A))

    for j in range (0, len(A)-1):
        P = pivoteo_partial(A)
        U = P @ U
        for i in range(j, len(A)):
          T[i][j] = -1 * (U[i][j]/U[j][j])

        U = T @ U
        L = T @ L

    return U,L

A=[[1, -8, 1],[-4, -2, 1],[1, -4, 1]]
U,L=resolver_PA_LU(A)


print("Matriz L")
print(L)

print("Matriz U")
print(U)
```
